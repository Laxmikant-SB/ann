```python
import numpy as np

# Define sigmoid activation function
def sigmoid(x):

    return 1 / (1 + np.exp(-x))

# Define derivative of sigmoid function
def sigmoid_derivative(x):

    return x * (1 - x)

# Input dataset
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Output dataset
y = np.array([
    [0],
    [1],
    [1],
    [0]
])

# Hyperparameters
learning_rate = 0.1
num_epochs = 100000

# Initialize weights randomly
hidden_weights = 2 * np.random.random((2, 2)) - 1

output_weights = 2 * np.random.random((2, 1)) - 1

# Training the neural network
for i in range(num_epochs):

    # =====================
    # Forward Propagation
    # =====================

    hidden_layer = sigmoid(
        np.dot(X, hidden_weights)
    )

    output_layer = sigmoid(
        np.dot(hidden_layer, output_weights)
    )

    # =====================
    # Back Propagation
    # =====================

    output_error = y - output_layer

    output_delta = (
        output_error *
        sigmoid_derivative(output_layer)
    )

    hidden_error = output_delta.dot(
        output_weights.T
    )

    hidden_delta = (
        hidden_error *
        sigmoid_derivative(hidden_layer)
    )

    # =====================
    # Update Weights
    # =====================

    output_weights += (
        hidden_layer.T.dot(output_delta)
        * learning_rate
    )

    hidden_weights += (
        X.T.dot(hidden_delta)
        * learning_rate
    )

# =====================
# Display Results
# =====================

print("Input:")
print(X)

print("\nPredicted Output:")
print(output_layer)
```
